Task 5 — Merging & Joining DataFrames


```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'customer_id': ['C01','C02','C03','C04','C05','C06','C07','C08','C09'],
    'customer_name': ['Ravi Kumar','Priya Sharma','Amit Singh','Sneha Patel',
                      'Karan Mehta','Divya Nair','Rohit Gupta','Anita Joshi','Vikram Das'],
    'tier': ['Gold','Silver','Platinum','Silver','Gold','Bronze','Silver','Platinum','Bronze'],
    'city': ['Mumbai','Delhi','Bangalore','Chennai','Hyderabad','Pune','Kolkata','Ahmedabad','Jaipur']
})

products = pd.DataFrame({
    'product': ['Laptop','Phone','Tablet','Headphones'],
    'brand': ['Dell','Samsung','Apple','Sony'],
    'cost_price': [60000, 18000, 28000, 2500],
    'warranty_years': [2, 1, 1, 1]
})

df = pd.read_csv('ecommerce_orders.csv')

print(f'Customers DataFrame:\n{customers}\n')
print(f'Products DataFrame:\n{products}\n')
print(f'Orders DataFrame:\n{df}\n')
```

```python
#a = pd.merge(df,customers, on='customer_id', how='left')
#print(f'Merged DataFrame:\n{a}\n')

b = pd.merge(df,customers, on='customer_id', how='inner')
print(f'Merged DataFrame inner join:\n{b}\n')
```

```python
# Q2. LEFT JOIN orders with customers — find orders with no customer info

a = pd.merge(df,customers, on='customer_id', how='left')
print(f'Merged DataFrame:\n{a}')

# order with no customer details will have NaN values in the merged DataFrame, indicating missing information.

# 19  2024-04-28   West       Sneha         0    Pending           NaN   
```

```python

# Q3. Merge orders with products on product name
#     → Calculate profit = amount - cost_price for each order
a = pd.merge(df, products, on='product', how='inner')
print(f'Merged DataFrame on product:\n{a}')

profit = a['amount'] - a['cost_price']
a['profit'] = profit
print(f'Merged DataFrame with profit:\n{a}')
```

```python
# Q4. After merging all 3: find profit by customer tier
merged_all = pd.merge(a, customers, on='customer_id', how='inner')
profit_by_tier = merged_all.groupby('tier')['profit'].sum().reset_index()
print(f'Profit by customer tier:\n{profit_by_tier}')
```

```python
# Q5. Find customers who have NEVER placed an order (use outer join + filter nulls)
```

```python
# since we want to find customers who have never placed an order, we will perform a left join of the customers DataFrame with the orders DataFrame. This way, all customers will be included in the result, and those who have never placed an order will have NaN values in the order-related columns.

# but in question it is mentioned to use outer join, so we will use outer join and then filter out the customers who have never placed an order by checking for null values in the order_id column.

customers_orders = customers.merge(df, on='customer_id', how='outer')
print(f'Customers merged with orders:\n{customers_orders}')
#customers_never_ordered = customers_orders[customers_orders['order_id'].isnull()]

# we can do like this:
#merged[merged['customer_name'].isnull()]  # no customer info
# earlier: result = customers_orders[customers_orders['order_id'].isna()]

result = customers_orders[customers_orders['customer_name'].isna()]
print(f'Customers who have never placed an order:\n{result}')
```

```python
customers_orders = customers.merge(df, on='customer_id', how='left')
most_rev_city= customers_orders.groupby('city')['amount'].sum().reset_index(name='total_amount_by_city').sort_values(by='total_amount_by_city', ascending=False).head().reset_index(drop=True)
print(f'Most revenue-generating city:\n{most_rev_city}')
```

```python
# Q7. Create a final "enriched_orders" dataframe with all info combined
#     Save it to "enriched_orders.csv"
enriched_orders = pd.merge(df, customers, on='customer_id', how='left')
enriched_orders = pd.merge(enriched_orders, products, on='product', how='left')
enriched_orders.to_csv("C:\\Users\\DEVANSH\\OneDrive\\Desktop\\Python-Data-Analysis\\Python_pandas\\enriched_orders.csv", index=False)
print(f'Enriched Orders DataFrame:\n{enriched_orders}')
```
