Task 3 — Filtering & Selection
Real-World Context: Answering specific business questions by slicing data.

Question
df = pd.read_csv('ecommerce_orders.csv')

 Answer these business questions using filters:
```
 Q1. Show all orders from the "North" region
 Q2. Find all orders with amount greater than ₹50,000
 Q3. Show all "Pending" orders from salesperson "Amit"
 Q4. Find orders where discount is 0 AND amount > 30000
 Q5. Show all orders for "Laptop" or "Tablet" (use isin)
 Q6. Find orders where amount is between ₹20,000 and ₹40,000
 Q7. Select only columns: order_id, product, amount, region
 Q8. Find the top 3 orders by amount in the "South" region
 Q9. Show orders where salesperson name starts with 'A' or 'P'
 Q10. Find orders placed in February 2024 (use order_date column)
 ```

````python
import pandas as pd
df = pd.read_csv('ecommerce_orders.csv')
print("=====ORIGINAL DATASET====")
print(df)
````

````python

n= df[df['region']== 'North']
print("=====FILTERED DATASET (region == 'North')====\n" , n) # or df[n] if n= df['region']== 'North'
````

````python
amt= df[df['amount']> 50000]
print("=====FILTERED DATASET (amount > 50000)==== \n" , amt) # or df[amt] if amt= df['amount']> 50000
````

````python
amt= df[(df['amount']> 50000) & (df['salesperson']=='Amit')]
print("=====FILTERED DATASET (amount > 50000 and salesperson == 'Amit')====\n" , amt) # or df[amt] if amt= df[(df['amount']> 50000) & (df['salesperson']=='Amit')]
````

````python
#4
amt= df[(df['amount']> 30000) & (df['discount']==0)]
print("=====FILTERED DATASET (amount > 30000 and discount == 0)====\n" , amt) # or df[amt] if amt= df[(df['amount']> 30000) & (df['discount']==0)]
````

````python

d= df[ df['product'].isin(['Laptop','Tablet'])]
print("=====FILTERED DATASET (product is either 'Laptop' or 'Tablet')==== \n" , d) # or df[d] if d= df[ df['product'].isin(['Laptop','Tablet'])]
````

````python
#6 order between amount1 and amount2 using df[''].between(amt1, amt2)
amt_range = df[df['amount'].between(20000, 40000)]
print("=====FILTERED DATASET (amount between 20000 and 40000)==== \n" , amt_range) # or df[amt_range] if amt_range = df['amount'].between(20000, 40000)
````

````python
#7
col = df[['order_id','product','amount','region']]
print("=====FILTERED DATASET (only order_id, product, amount, region columns)==== \n" , col) # or df[col] if col = df[['order_id','product','amount','region']]
````

````python
top=df.sort_values(by='amount', ascending=False).head(3) 
print("=====TOP 3 ORDERS BY AMOUNT IN DESCENDING ORDER==== \n" , top) # or df[top] if top=df.sort_values(by='amount', ascending=False).head(3)
````

````python
# Q9. Show orders where salesperson name starts with 'A' or 'P'
salesperson_filter = df[df['salesperson'].str.startswith(('A', 'P'))]
print("=====ORDERS WHERE SALESPERSON NAME STARTS WITH 'A' OR 'P'==== \n" , salesperson_filter) # or df[salesperson_filter] if salesperson_filter = df['salesperson'].str.startswith(('A', 'P'))
````

````python
# Q10. Find orders placed in February 2024 (use order_date column) using datetime functions
df['order_date'] = pd.to_datetime(df['order_date'])
february_2024_orders = df[(df['order_date'].dt.month == 2) & (df['order_date'].dt.year == 2024)]
print("=====ORDERS PLACED IN FEBRUARY 2024==== \n" , february_2024_orders) # or df[february_2024_orders] if february_2024_orders = df[(df['order_date'].dt.month == 2) & (df['order_date'].dt.year == 2024)]
````
