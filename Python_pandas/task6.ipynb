Task 6 — Pivot Tables


1. What is a Pivot Table (in Pandas)?

👉 A pivot table is just:

**A way to summarize data by grouping + aggregating + reshaping**

It answers questions like:

- Total sales by region?
- Avg revenue per product?
- Count of orders per customer?

2. Core Idea (VERY IMPORTANT)

Pivot = 3 things combined:

- Group by (rows)
- Split by (columns)
- Aggregate (values)

**Mental Model**

Think like this:

```
Rows → grouping level
Columns → categories to spread
Values → what to calculate
```

3. Basic Syntax

pd.pivot_table(
    data=df,
    index='row_column',
    columns='column_category',
    values='value_column',
    aggfunc='sum'
)

**Fill Missing Values** : Fill missing value to a given value
 `pd.pivot_table(
    df,
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum',
    fill_value=0
)`

**Margins (Grand Total)** :  provides grand total 

👉 Adds:

* Row total
* Column total

`pd.pivot_table(
    df,
    index='Region',
    columns='Product',
    values='Sales',
    aggfunc='sum',
    margins=True
) `

````python
import pandas as pd
````

````python
df = pd.read_csv('ecommerce_orders.csv')
print(f'Orders DataFrame:\n{df}\n')
````

````python
# Q1. Create a pivot: Rows = Region, Columns = Product, Values = Total Amount

sum= pd.pivot_table( df, index='region', columns= 'product', values= 'amount', aggfunc='sum', fill_value=0)
print(f'Pivot Table (Total Amount by Region and Product):\n{sum}')
````

````python
print(df.groupby(['product', 'region'])['amount'].sum())
````

````python
# Q2. Pivot: Salesperson vs Region showing COUNT of orders
````

````python
d= pd.pivot_table(df, index='salesperson', columns='region', values='order_id', aggfunc='count', fill_value=0)
print(f'Pivot Table (Count of Orders by Salesperson and Region):\n{d}')
````

````python
# Q3. Pivot: Product vs Status showing total Amount + add row/column totals (margins=True)
````

````python
product_status_pivot = pd.pivot_table(df, index='product', columns='status', values='amount', aggfunc='sum', margins=True, fill_value=0)
print(f'Pivot Table (Total Amount by Product and Status with Totals):\n{product_status_pivot}')
````

````python
# Q4. Multi-value pivot: Region vs Product showing both SUM of amount AND COUNT of orders

multi_value = pd.pivot_table(df, index= 'region', columns='product', values= ['amount', 'order_id'], aggfunc={'amount':'sum','order_id':'count'}, fill_value=0)
print(f'Pivot Table (Total Amount and Count of Orders by Region and Product):\n{multi_value}')
````

````python
# Q5. Create a pivot showing average discount per Salesperson per Region
````

````python
avg_disc= pd.pivot_table(df, index='salesperson', columns='region', values='discount', aggfunc='mean', fill_value=0)
print(f'Pivot Table (Average Discount by Salesperson and Region):\n{avg_disc}')
````

````python
# Q6. From Q1 pivot — find which product has highest sales in each region
highest_sales = sum.idxmax(axis=1)
print(f'Product with Highest Sales in Each Region:\n{highest_sales}')
````

````python
# Q7. Flatten the multi-index columns from Q4 into readable names
#     e.g., ('amount','sum') → 'amount_sum'
````

````python
#multi_value.columns= ['_'.join(col).strip() for col in multi_value.columns.values]

metric_map = {'amount': 'amount', 'order_id': 'order_count'}
multi_value.columns = [f'{metric_map[col[0]]}_{col[1]}' for col in multi_value.columns.values]

print(f'Flattened Pivot Table:\n{multi_value}')
````
