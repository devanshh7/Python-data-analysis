```python
import pandas as pd
df = pd.read_csv('ecommerce_orders.csv')
print(df.head())
```

```python
# Q1. Total sales amount by Region

reg_sales= df.groupby('region')['amount'].sum()
reg_sales = reg_sales.reset_index(name='total_sales')
print(reg_sales)
```

```python
# Q2. Average order amount by Product

avd_ord= df.groupby('product')['amount'].mean()
print('Average order amount by Product:\n', avd_ord)
```

```python
# Q3. Total sales AND order count by Salesperson

sales_count= df.groupby('salesperson')['amount'].agg(['sum', 'count'])
print('Total sales AND order count by Salesperson:\n', sales_count)
```

```python
# Q4. Maximum and minimum order amount by Category

m= df.groupby('category')['amount'].agg(['min','max'])
print('Minimum and Maximum sales amount by Category:\n', m)
```

```python
# Q5. Total quantity sold by Product (sorted descending)

f = df.groupby('product')['quantity'].sum().sort_values(ascending=False)
print(f)
```

```python
# Q6. How many orders per Status per Region? (multi-level groupby)

d = df.groupby('region')['status'].value_counts()
print(d)
```

```python
# Q7. What % of total sales does each Region contribute?
#     (hint: divide group sum by total sum * 100)
#want to add percentage sign and how can we give a column name like percent_by_region

total_sales = df['amount'].sum()
reg_sales_percent = (reg_sales / total_sales) * 100
reg_sales_percent = reg_sales_percent.reset_index(name='percent_by_region (%)').round(2)
print('Percentage of total sales by Region:\n', reg_sales_percent)
```

```python

s=df.groupby('salesperson')['discount'].agg(['mean', 'min', 'max'])
print('Average, Minimum and Maximum discount by Salesperson:\n', s)
```

```python
# Q9. Total revenue per month (group by month of order_date)
df['order_date'] = pd.to_datetime(df['order_date'])

df['month']= df['order_date'].dt.month
rpm= df.groupby('month')['amount'].sum().reset_index(name='total_revenue')
print('Total revenue per month:\n', rpm)
```

```python
a=df.groupby(['region', 'salesperson'])['amount'].sum().reset_index(name='total_revenue')
max_total_revenue = a.groupby('region')['total_revenue'].max()
print(a)
print('Maximum total revenue across all regions and salespersons:\n', max_total_revenue)
```

```python
# Q10. Find the salesperson with highest total sales in each region
```

```python
a= df.groupby(['region', 'salesperson'])['amount'].sum().reset_index(name='total_revenue')
print(a)

#b= a.groupby('region')['total_revenue'].max().reset_index(name='max_revenue')
result = a.loc[a.groupby('region')['total_revenue'].idxmax()]

#result = a.merge(b, on=['region', 'total_revenue'])
print(f'\nMaximum total revenue in each region by Salesperson:\n{result}')
```
